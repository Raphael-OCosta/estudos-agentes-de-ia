#### 🧪 Exercício: Teste A/B de dois prompts

O prompt A resolveu 412 de 500 conversas e o prompt B, 441 de 500. Calcule a diferença, a estatística z para duas proporções e o p-valor. Vale trocar para o B?

```python
from math import sqrt, erf
a, na, b, nb = 412, 500, 441, 500
pa, pb = a / na, b / nb; p = (a + b) / (na + nb)
z = (pb - pa) / sqrt(p * (1 - p) * (1 / na + 1 / nb))
p_valor = 2 * (1 - (1 + erf(abs(z) / sqrt(2))) / 2)
print(f"diferença {pb - pa:.3f} | z = {z:.2f} | p = {p_valor:.4f}")
```
